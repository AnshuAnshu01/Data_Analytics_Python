```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("netflix_customer.csv")
```

```python
plt.figure(figsize=(18,10))

plt.subplot(2,4,1)
sns.countplot(data=df, x="gender", hue="gender", palette="Blues", edgecolor="black", legend=False)
plt.xlabel("Gender")
plt.ylabel("count")
plt.title("Distribution of Gender")

plt.subplot(2,4,2)
sns.countplot(data=df, x="subscription_type", hue="subscription_type", palette="Blues", edgecolor="black", legend=False)
plt.xlabel("Subscription Type")
plt.ylabel("count")
plt.title("Distribution of Subscription Type")

plt.subplot(2,4,3)
sns.countplot(data=df, x="region", hue="region", palette="Blues", edgecolor="black", legend=False)
plt.xlabel("Region")
plt.ylabel("count")
plt.title("Distribution of Region")
plt.xticks(rotation=25)

plt.subplot(2,4,4)
sns.countplot(data=df, x="device", hue="device", palette="Blues", edgecolor="black", legend=False)
plt.xlabel("Device")
plt.ylabel("count")
plt.title("Distribution of Device")

plt.subplot(2,4,5)
sns.countplot(data=df, x="payment_method", hue="payment_method", palette="Blues", edgecolor="black", legend=False)
plt.xlabel("Payment Method")
plt.ylabel("count")
plt.title("Distribution of Payment Method")
plt.xticks(rotation=25)

plt.subplot(2,4,6)
sns.countplot(data=df, x="favorite_genre", hue="favorite_genre", palette="Blues", edgecolor="black", legend=False)
plt.xlabel("Favorite Genre")
plt.ylabel("count")
plt.title("Distribution of Favorite Genre")
plt.xticks(rotation=25)

plt.subplot(2,4,7)
sns.countplot(data=df, x="monthly_fee", hue="monthly_fee", palette="Blues", edgecolor="black", legend=False)
plt.xlabel("Monthly Fee")
plt.ylabel("count")
plt.title("Distribution of Monthly Fee")

plt.subplot(2,4,8)
sns.countplot(data=df, x="number_of_profiles", hue="number_of_profiles", palette="Blues", edgecolor="black", legend=False)
plt.xlabel("Number of Profiles")
plt.ylabel("count")
plt.title("Distribution of Number of Profiles")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
sns.countplot(data=df, x="churned", palette="Blues", edgecolor="black")
plt.xlabel("Churned")
plt.ylabel("count")
plt.title("Distribution of Customer Churn")
plt.subplot(1,2,2)
plt.pie(df["churned"].value_counts(), labels=df["churned"].value_counts().index, autopct="%0.2f%%", colors=sns.color_palette("Blues"))
plt.title("Customer Churn Composition")
plt.tight_layout()
plt.show()
```

```python
plt.pie(df["subscription_type"].value_counts(),
        labels=df["subscription_type"].value_counts().index,
        autopct="%0.2f%%",
        colors=["#808080", "#e50914", "#b20710"])

plt.title("Customers by Subscription Type")
plt.tight_layout()
plt.show()
```

```python
plt.pie(df["gender"].value_counts(),
        labels=df["gender"].value_counts().index,
        autopct="%0.2f%%",
        colors=sns.color_palette(["#00FF00", "#E50914", "#564d4d"]))

plt.title("Customers by Gender")
plt.tight_layout()
plt.show()
```

```python
device_count = df["device"].value_counts()

plt.figure(figsize=(10, 6))
plt.barh(device_count.index, device_count.values, color="#E50914", edgecolor="black")

plt.title("Customers by Device Used")
plt.xlabel("Number of Customers")
plt.ylabel("Device")
plt.tight_layout()
plt.show()
```

```python
payment_count = df["payment_method"].value_counts()

plt.figure(figsize=(10, 6))
plt.barh(payment_count.index, payment_count.values,
         color=sns.color_palette("Reds", len(payment_count)),
         edgecolor="black")

plt.xlabel("Number of Customers")
plt.ylabel("Payment Method")
plt.title("Customers by Payment Method")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
