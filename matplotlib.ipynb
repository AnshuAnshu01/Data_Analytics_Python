```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10, 20, 25, 35, 45]
plt.plot(x,y)
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.plot(x,y, color='red', marker='o', linestyle='--', linewidth=2, markersize=12)

plt.title('Sample Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
x = [1,2,3,4,5]
y1 = [10, 20, 25, 35, 45]
y2 = [22, 30, 15, 45, 55]
plt.plot(x,y1, label="sales 2025")
plt.plot(x,y2, label="sales 2026")

plt.title("YOY Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [10, 20, 25, 35, 45]

plt.bar(x,y ,color="pink")
plt.title("Bar Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# Histogram use for distribution analysis
import random
data=[random.randint(0,10) for i in range(100)]
plt.hist(data, bins=10 ,color="black")
plt.title("Histogram")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D', 'E']
values = [10, 20, 25, 35, 45]

plt.pie(values, labels=categories, autopct='%1.1f%%',startangle=120)
plt.title('Pie Chart')
plt.show()
```

```python
import random
data=[random.randint(0,10) for i in range(50)]
data2=[random.randint(1,20) for i in range(50)]
data3=[random.randint(0,30) for i in range(50)]
data4=[random.randint(1,40) for i in range(50)]
plt.scatter(data,data2,color="red")
plt.scatter(data3,data4,color="black")
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

```python
#subplots
# data 1 bar chart
categories = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
values = [10, 20, 25, 35, 45]
#data 2 - scaatter plot
y1=[10,20,30,40,50]
y2=[13,25,38,47,55]

plt.figure(figsize=(15,5))

#first plot bar chart
plt.subplot(1,3,1) #row ,column ,position
plt.bar(categories,values,color="pink")
plt.title("Daily Sales")
plt.xlabel("Days")
plt.ylabel("Sales")

#second plot scaatter plot
plt.subplot(1,3,2)
plt.scatter(y1,y2,color="black")
plt.title("Users Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)

#3rd
plt.subplot(1,3,3)
plt.pie(values, labels=categories, autopct='%1.1f%%',startangle=120)
plt.title('Pie Chart')



plt.show()

```

```python
import pandas as pd
df=pd.read_csv("data4.csv")
plt.figure(figsize=(15,5))
plt.subplot(1,2,1)
plt.bar(df['channel_info'],df['influence_score'] , color="pink")
plt.title('Bar Chart')
plt.subplot(1,2,2)
plt.pie(df['influence_score'], labels=df['channel_info'], autopct='%1.1f%%',startangle=90)
plt.title('Pie Chart')
plt.show()
```

```python
#Line plot of influence score against rank.
plt.figure(figsize=(8, 4))
plt.plot(df["rank"], df["influence_score"], color="purple")
plt.title("Influence Score by Rank")
plt.xlabel("Rank"); plt.ylabel("Influence score")
plt.grid()
plt.show()
```

```python
df=pd.read_csv("data4.csv")
plt.figure(figsize=(8, 4))
plt.hist(df["60_day_eng_rate"], bins=15, color="skyblue", edgecolor="black")
plt.axvline(df["60_day_eng_rate"].mean(), color="red", linestyle="--", label="Mean")
plt.xlabel("Engagement rate (%)"); plt.ylabel("Count")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(df["followers"] / 1000000, df["avg_likes"] / 1000000, color="red")
plt.plot(df["followers"] / 1000000, df["avg_likes"] / 1000000, color="blue", linestyle="-")
plt.grid()
plt.xlabel("Followers (M)")
plt.ylabel("Avg likes (M)")
plt.title("Followers vs Avg Likes")
plt.show()
```

```python
top10 = df.nlargest(10,"followers")
plt.figure(figsize=(8, 8))
plt.pie(top10["followers"],labels=top10['channel_info'],autopct="%1.1f%%",startangle=90)
plt.title("Top 10 Accounts by Followers")
plt.show()
```

```python
cnt = df["country"].value_counts()
top5 = cnt.head()
top5["Others"] = cnt.iloc[5:].sum()
plt.figure(figsize=(6, 6))
plt.pie(top5, labels=top5.index, autopct="%1.1f%%", startangle=90)
plt.title("Influencers by Country")
plt.show()
```

```python
cnt = df["country"].value_counts()
others = cnt.iloc[5:]
plt.figure(figsize=(10, 6))
plt.bar(others.index, others.values, color="orange")
plt.xlabel("Country")
plt.ylabel("Number of Influencers")
plt.title("Influencers from Other Countries")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
