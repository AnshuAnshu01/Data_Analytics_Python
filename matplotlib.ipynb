```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10, 20, 25, 35, 45]
plt.plot(x,y)
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.plot(x,y, color='red', marker='o', linestyle='--', linewidth=2, markersize=12)

plt.title('Sample Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```
