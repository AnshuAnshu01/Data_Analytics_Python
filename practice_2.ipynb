```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from mpl_toolkits import mplot3d
```

```python
ax=plt.axes(projection='3d')
x_data=np.random.randint(150, size=(1000))
y_data=np.random.randint(150, size=(1000))
z_data=np.random.randint(150, size=(1000))

ax.scatter(x_data, y_data, z_data, c=z_data)
plt.show()
```

```python
ax=plt.axes(projection='3d')
x=np.random.randint(5, size=(3))
y=np.random.randint(5, size=(3))
z=np.random.randint(5, size=(3))

dx = np.ones(3)
dy = np.ones(3)
dz = np.ones(3)

ax.bar3d(x, y, z,dx,dy,dz)
plt.show()
```
