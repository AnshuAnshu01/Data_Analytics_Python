# Simple Guide to Data Visualization with Seaborn and Plotly

In this notebook, we will learn how to make simple data visualizations. We'll explore datasets, ask straightforward questions, and explain exactly what the code does and what the outputs show.

## Section 1: Setup and Libraries
We first import our essential plotting and data libraries: `matplotlib.pyplot` for managing figures, `seaborn` for easy-to-use statistical plots, and `plotly` for interactive data.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# QUESTION: How do we import visualization libraries and configure the default style?
# EXPLANATION: We import plotting libraries and tell Seaborn to use its standard darkgrid design style.
sns.set_theme(style="darkgrid")
```

## Section 2: Loading Our Data
**Question:** How do we load a simple dataset about restaurant tips and see what is inside?

**What we do:** We use Seaborn's built-in `load_dataset` function to fetch the 'tips' dataset, then display its first few rows.

```python
# QUESTION: How do we load a simple restaurant dataset and look at the first few records?
# EXPLANATION: We use load_dataset to load the 'tips' data and display the top 5 rows.
tips = sns.load_dataset('tips')
display(tips.head())
print(tips.info())

# WHAT THE OUTPUT SHOWS:
# A clean table detailing the total bill, tip, server's gender, smoker status, day, time, and table size.
```

## Section 3: Scatter Plots (Relationships)
**Question:** Do people who pay higher total bills also leave larger tips?

**What we do:** We create a simple scatter plot putting `total_bill` on the horizontal X-axis and `tip` on the vertical Y-axis.

```python
# QUESTION: Do higher total bills lead to larger tips?
# EXPLANATION: We draw a scatter plot showing 'total_bill' on the X-axis and 'tip' on the Y-axis.
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time')
plt.title('Relationship between Total Bill and Tip Amount')
plt.show()

# WHAT THE OUTPUT SHOWS:
# Individual dots representing each bill. The upward trend of dots reveals a positive correlation: larger bills generally yield higher tips.
```

## Section 4: Distribution Plots
**Question:** What are the most common total bill values in our dataset?

**What we do:** We use a histogram (`histplot`) to group the total bill amounts into bins and count how many bills fall into each range.

```python
# QUESTION: What are the most common total bill values in our dataset?
# EXPLANATION: We use a histogram (histplot) to count how frequently different bill ranges occur.
plt.figure(figsize=(8, 5))
sns.histplot(data=tips, x='total_bill', kde=True, color='purple')
plt.title('Distribution of Total Bills')
plt.show()

# WHAT THE OUTPUT SHOWS:
# The tall bars show that most bills fall between $10 and $20. The smooth purple line shows the density trend.
```

## Section 5: Categorical Plots
**Question:** On which days of the week do customers pay the highest bills?

**What we do:** We use a boxplot to summarize the spread of total bills for each day (Thursday through Sunday).

```python
# QUESTION: On which days of the week do customers pay the highest bills?
# EXPLANATION: We construct boxplots divided by day of the week to analyze the middle spread and median values.
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x='day', y='total_bill', hue='day', palette='Set2')
plt.title('Total Bills Divided by Day of Week')
plt.show()

# WHAT THE OUTPUT SHOWS:
# The boxes represent where the middle 50% of bills sit. Weekend boxes (Sat/Sun) are placed higher up than weekday boxes, showing people spend more on weekends.
```

## Section 6: Regression Plots
**Question:** Can we draw a straight trend line to see the direct relationship between bill sizes and tips?

**What we do:** We use `regplot` to draw both the scatter points and a calculated line of best fit.

```python
# QUESTION: Can we plot a straight line indicating the exact overall trend between bills and tips?
# EXPLANATION: We use a regression plot (regplot) which automatically calculates and draws a line of best fit over the scattered points.
plt.figure(figsize=(8, 5))
sns.regplot(data=tips, x='total_bill', y='tip', scatter_kws={'color': 'blue'}, line_kws={'color': 'red'})
plt.title('Trend Line: Tip vs Total Bill')
plt.show()

# WHAT THE OUTPUT SHOWS:
# A red trend line moving up towards the right, proving mathematically that larger bills lead to higher tips.
```
