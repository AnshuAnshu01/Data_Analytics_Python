Array Reshaping and Flattening

```python
arr6 = np.array([1,2,3,4,5,6])
```

```python
respaced_arr = arr6.reshape(6,1)
respaced_arr2 = arr6.reshape(3,2)
print(respaced_arr)
print(respaced_arr2)
```

```python
print(respaced_arr2.flatten())  #2d to 1d conversion
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.vstack((a,b))) # output array is 2D
print(np.hstack((a,b))) # output array is 1D
```

```python
c = np.array([[1,2,3], [4,5,6]])
print(c)
```

```python
hsplit = np.hsplit(c,3)
print(hsplit)  # horizontal split 

for s in hsplit:
    print(s)
```

```python
vsplit = np.vsplit(c,2) # vertical split
print(vsplit)  

for s in vsplit:
    print(s)
```

```python
A = np.array([10,20,40,-30])
print(a+10) 
print(a-30) 
print(a*2) 
print(a/10)
```

```python
B=np.array([ 4,9,144,12
])
print(np.sqrt(B))
print(np.square(B))
print(np.log(B))
print(np.sin(B))
```

```python
aa=np.array([[1,2,3],[4,5,6]])
bb=np.array([[7,8,9],[10,11,12]])
print(np.add(aa,bb))
print(np.subtract(aa,bb))
print(np.multiply(aa,bb))
print(np.divide(aa,bb))
print(np.dot(aa,bb.T))
print(np.dot(bb,aa.T))
print(np.dot(A,B))
```

```python
t = np.array([[1,2,3],
              [4,5,6]])

print(t)    
print(t.T)  # transposed
```
