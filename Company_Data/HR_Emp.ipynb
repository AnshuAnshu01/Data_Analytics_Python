```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("HR-Employee-Attrition.csv")
df_l=["Age","Attrition","Department","BusinessTravel","DistanceFromHome","Gender","HourlyRate","JobRole","MaritalStatus","MonthlyRate"]
df1=df[df_l]
```

```python
df1.info()
```

```python
df1.head()
```

```python
gender_count = df1["Gender"].value_counts()

plt.pie(
    gender_count.values,
    labels=gender_count.index,
    autopct="%1.1f%%",
    colors=["orange", "pink"],
    startangle=90
)

plt.title("Male and Female Distribution")
plt.show()
```

```python
plt.hist(df["Age"], bins=10, color="skyblue", edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

```python
department_count = df["Department"].value_counts()

plt.bar(department_count.index, department_count.values,
        color="lightgreen", edgecolor="black")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.title("Employees by Department")
plt.xticks(rotation=15)
plt.show()
```

```python
attrition_count = df["Attrition"].value_counts()

plt.pie(attrition_count.values, labels=attrition_count.index,
        autopct="%1.1f%%", colors=["lightblue", "tomato"],
        startangle=90)
plt.title("Employee Attrition Distribution")
plt.show()
```

```python
travel_count = df["BusinessTravel"].value_counts()

plt.bar(travel_count.index, travel_count.values,
        color="orange", edgecolor="black")
plt.xlabel("Business Travel")
plt.ylabel("Number of Employees")
plt.title("Employees by Business Travel")
plt.xticks(rotation=15)
plt.show()
```

```python
plt.hist(df["DistanceFromHome"], bins=10,
         color="lightcoral", edgecolor="black")
plt.xlabel("Distance From Home")
plt.ylabel("Frequency")
plt.title("Employee Distance From Home")
plt.show()
```

```python
plt.boxplot(df["HourlyRate"])
plt.ylabel("Hourly Rate")
plt.title("Distribution of Hourly Rates")
plt.show()
```

```python
job_count = df["JobRole"].value_counts()

plt.bar(job_count.index, job_count.values,
        color="mediumpurple", edgecolor="black")
plt.xlabel("Job Role")
plt.ylabel("Number of Employees")
plt.title("Employees by Job Role")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
marital_count = df["MaritalStatus"].value_counts()

plt.pie(marital_count.values, labels=marital_count.index,
        autopct="%1.1f%%", startangle=90)
plt.title("Employee Marital Status")
plt.show()
```

```python
plt.hist(df["MonthlyRate"], bins=15,
         color="gold", edgecolor="black")
plt.xlabel("Monthly Rate")
plt.ylabel("Frequency")
plt.title("Distribution of Monthly Rates")
plt.show()
```

```python
avg_age = df.groupby("Department")["Age"].mean()

plt.bar(avg_age.index, avg_age.values,
        color="teal", edgecolor="black")
plt.xlabel("Department")
plt.ylabel("Average Age")
plt.title("Average Employee Age by Department")
plt.show()
```

```python
gender_department = pd.crosstab(df["Department"], df["Gender"])

gender_department.plot(kind="bar", color=["orange", "pink"],
                       edgecolor="black")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.title("Gender Distribution by Department")
plt.xticks(rotation=0)
plt.legend(title="Gender")
plt.tight_layout()
plt.show()
```

```python
left_employees = df[df["Attrition"] == "Yes"]
left_count = left_employees["Department"].value_counts()

plt.bar(left_count.index, left_count.values,
        color="tomato", edgecolor="black")
plt.xlabel("Department")
plt.ylabel("Employees Who Left")
plt.title("Employee Attrition by Department")
plt.show()
```

```python
gender_attrition = pd.crosstab(df["Gender"], df["Attrition"])

gender_attrition.plot(
    kind="bar",
    color=["skyblue", "orange"],
    edgecolor="black",
    figsize=(8, 5)
)

plt.xlabel("Gender")
plt.ylabel("Employee Count")
plt.title("Attrition by Gender")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.show()
```
