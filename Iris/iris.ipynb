**Question 23:** Use the `.apply()` method or custom function to create a column `size_category` which labels a flower as 'Large' if its `petal_length` is >= 4.5, and 'Small' otherwise.

### Part 2: Advanced Visualizations (Questions 26-35)

**Question 26:** Plot a histogram showing the distribution of `sepal_len` with a kernel density estimate (KDE) curve overlaid on top.

```python
# Write your code here
plt.figure(figsize=(8, 5))

sns.histplot(df["sepal_len"], kde=True)

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")

plt.show()
```

**Question 27:** Create a grouped bar chart showing the mean values of the 4 numeric features (`sepal_len`, `sepal_wid`, `petal_length`, `petal_width`) grouped by each `species`.

```python

mean_data = df.groupby("species")[
    ["sepal_len", "sepal_wid", "petal_length", "petal_width"]
].mean()


x = np.arange(len(mean_data.index))
width = 0.2

plt.figure(figsize=(10, 6))


plt.bar(x - 1.5*width, mean_data["sepal_len"], width, label="Sepal Length")
plt.bar(x - 0.5*width, mean_data["sepal_wid"], width, label="Sepal Width")
plt.bar(x + 0.5*width, mean_data["petal_length"], width, label="Petal Length")
plt.bar(x + 1.5*width, mean_data["petal_width"], width, label="Petal Width")


plt.xlabel("Species")
plt.ylabel("Mean Values")
plt.title("Mean Features Comparison by Species")

plt.xticks(x, mean_data.index)
plt.legend()

plt.show()
```

**Question 28:** Draw a bar plot showing the average `sepal_area` for each `species` with standard error bars.

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df,x="species",y="sepal_area")

plt.title("Average Sepal Area by Species")
plt.xlabel("Species")
plt.ylabel("Average Sepal Area")

plt.show()
```

**Question 29:** Plot a joint distribution plot (jointplot) showing the relationship between `petal_length` and `petal_width` with marginal histograms, colored by `species`.

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 10))

sns.scatterplot(data=df, x="petal_length", y="petal_width",
                hue="species", ax=ax[0, 0])
ax[0, 0].set_title("Petal Length vs Petal Width")

sns.scatterplot(data=df, x="sepal_len", y="sepal_wid",
                hue="species", ax=ax[0, 1])
ax[0, 1].set_title("Sepal Length vs Sepal Width")

sns.scatterplot(data=df, x="sepal_len", y="petal_length",
                hue="species", ax=ax[1, 0])
ax[1, 0].set_title("Sepal Length vs Petal Length")

sns.scatterplot(data=df, x="sepal_wid", y="petal_width",
                hue="species", ax=ax[1, 1])
ax[1, 1].set_title("Sepal Width vs Petal Width")

plt.tight_layout()
plt.show()
```

**Question 30:** Create a stacked bar chart showing the count of flowers in each `species` classified under each `size_category` (Large vs Small).

```python
# Write your code here
size_count = pd.crosstab(df["species"], df["size_category"])

size_count.plot(kind="bar", stacked=True, figsize=(8, 5))

plt.title("Species Count by Size Category")
plt.xlabel("Species")
plt.ylabel("Count")
plt.legend(title="Size Category")
plt.show()
```

**Question 31:** Plot a strip plot (using seaborn) overlaid on a box plot to visualize the distribution of `petal_length` for each `species` without hiding individual data points.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="species", y="petal_length")

sns.stripplot(data=df, x="species", y="petal_length",
              color="black", alpha=0.5, jitter=True)

plt.title("Petal Length Distribution by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")

plt.show()
```

**Question 32:** Create a 2D density plot (hexbin plot) showing the density of observations across `sepal_len` and `sepal_wid`.

```python
# Write your code here
plt.figure(figsize=(8, 6))

plt.hexbin(df["sepal_len"], df["sepal_wid"], gridsize=20, cmap="viridis")

plt.colorbar(label="Count")

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Sepal Length vs Sepal Width - Hexbin Plot")

plt.show()
```

**Question 33:** Generate a line plot showing the sorted values of `sepal_len` for each species separately on the same axis to compare their cumulative-style distribution profiles.

```python
plt.figure(figsize=(10, 6))

setosa = df[df["species"] == "setosa"]["sepal_len"].sort_values().reset_index(drop=True)
versicolor = df[df["species"] == "versicolor"]["sepal_len"].sort_values().reset_index(drop=True)
virginica = df[df["species"] == "virginica"]["sepal_len"].sort_values().reset_index(drop=True)

plt.plot(setosa, label="Setosa", marker="o")
plt.plot(versicolor, label="Versicolor", marker="o")
plt.plot(virginica, label="Virginica", marker="o")

plt.xlabel("Sample Index (Sorted)")
plt.ylabel("Sepal Length")
plt.title("Sorted Sepal Length Comparison by Species")

plt.legend()
plt.show()
```

**Question 34:** Use matplotlib's `subplot` structure to plot two side-by-side scatter plots: the left comparing `sepal_len` vs `sepal_wid`, and the right comparing `petal_length` vs `petal_width`, both styled with a dark theme background style.

**Question 35:** Draw an Andrews Curves plot or a Parallel Coordinates plot grouped by `species` using pandas visualization tools to visualize multi-dimensional feature patterns.
